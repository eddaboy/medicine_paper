```python
import hashlib
import json
import math
import os
import re
import time
from pathlib import Path
from typing import Any, Dict, List
import numpy as np
import pandas as pd
from tqdm.auto import tqdm
try:
    from openai import OpenAI
except ImportError:
    OpenAI = None
pd.set_option('display.float_format', '{:.3f}'.format)
pd.set_option('display.max_colwidth', 90)
```

```python
ROOT = Path('../../experiments/clinical_protocols').resolve()
assert (ROOT / 'delphi_panels').exists(), f'set ROOT manually; got {ROOT}'
DESIGN = 'noroles'
ARMS = {'delphi': ROOT / f'delphi_panels/heterogeneous_gpt120_organizer_complete/llm_as_judge/delphi_base_{DESIGN}_judge.csv'}
BASELINE_ARMS = ['baseline_gpt', 'baseline_qwen', 'baseline_deepseekv4']
SUP_SCORES_CSV = Path('sup/geval_scores.csv')
JOIN_KEY = 'Vignette'
MODEL = 'gpt-4.1-mini-2025-04-14'
API_KEY = ''
SCALE_LO, SCALE_HI = (1, 5)
TEMPERATURE_SCORE = 0.0
MAX_TOKENS_SCORE = 5
TOP_LOGPROBS = 20
TEMPERATURE_STEPS = 0.3
MAX_TOKENS_STEPS = 900
REQUEST_TIMEOUT = 60
BASE_SLEEP_SECONDS = 2
MAX_RETRIES = 6
RUN = True
SAMPLE_N = None
STEPS_JSON = Path('geval_steps.json')
OUT_CSV = Path('geval_scores_framework_only.csv')
```

```python
def read_csv(path):
    d = pd.read_csv(path, sep=';', low_memory=False)
    d.columns = [c.lstrip('\ufeff') for c in d.columns]
    return d

def strip_author(text):
    if not text:
        return ''
    return re.split('(?im)^\\s*Protocol Author\\s*:', str(text), maxsplit=1)[0].rstrip()

def n_words(text):
    return len(re.findall('\\w+', str(text)))
frames = []
for arm, path in ARMS.items():
    d = read_csv(path)
    missing = [c for c in (JOIN_KEY, 'case_text', 'winner_protocol') if c not in d.columns]
    if missing:
        raise ValueError(f'{arm}: нет колонок {missing}')
    d = d[[JOIN_KEY, 'case_text', 'winner_protocol']].copy()
    d = d[d['winner_protocol'].notna()]
    d['arm'] = arm
    d['protocol'] = d['winner_protocol'].map(strip_author)
    d['words'] = d['protocol'].map(n_words)
    frames.append(d[[JOIN_KEY, 'arm', 'case_text', 'protocol', 'words']])
protocols = pd.concat(frames, ignore_index=True)
if SAMPLE_N:
    keep = protocols[JOIN_KEY].drop_duplicates().head(SAMPLE_N)
    protocols = protocols[protocols[JOIN_KEY].isin(keep)].reset_index(drop=True)
print(f'рук: {protocols.arm.nunique()}   кейсов: {protocols[JOIN_KEY].nunique()}   протоколов: {len(protocols)}')
protocols.groupby('arm').agg(n=('protocol', 'size'), mean_words=('words', 'mean')).round(0)
```

```python
judge_client = None
llm_cache: Dict[str, Any] = {}

def _cache_key(*parts):
    raw = json.dumps({'model': MODEL, 'parts': list(parts)}, sort_keys=True, ensure_ascii=False)
    return hashlib.sha256(raw.encode('utf-8')).hexdigest()

def _client():
    global judge_client
    if judge_client is None:
        if not API_KEY:
            raise RuntimeError('OPENAI_API_KEY не задан в окружении')
        judge_client = OpenAI(api_key=API_KEY)
    return judge_client

def call_text(system_text, user_text, temperature, max_tokens):
    key = _cache_key('text', system_text, user_text, temperature, max_tokens)
    if key in llm_cache:
        return llm_cache[key]
    last = ''
    for attempt in range(MAX_RETRIES):
        try:
            r = _client().chat.completions.create(model=MODEL, messages=[{'role': 'system', 'content': system_text}, {'role': 'user', 'content': user_text}], temperature=temperature, max_tokens=max_tokens, timeout=REQUEST_TIMEOUT)
            out = r.choices[0].message.content.strip()
            llm_cache[key] = out
            return out
        except Exception as e:
            last = str(e)
            time.sleep(BASE_SLEEP_SECONDS * (attempt + 1))
    raise RuntimeError(f'call_text failed after {MAX_RETRIES} retries: {last}')

def call_score(prompt_text):
    key = _cache_key('score', prompt_text)
    if key in llm_cache:
        return llm_cache[key]
    last = ''
    for attempt in range(MAX_RETRIES):
        try:
            r = _client().chat.completions.create(model=MODEL, messages=[{'role': 'system', 'content': prompt_text}], temperature=TEMPERATURE_SCORE, max_tokens=MAX_TOKENS_SCORE, logprobs=True, top_logprobs=TOP_LOGPROBS, timeout=REQUEST_TIMEOUT)
            dist = _dist_from_logprobs(r)
            out = {'dist': dist, 'text': (r.choices[0].message.content or '').strip()}
            llm_cache[key] = out
            return out
        except Exception as e:
            last = str(e)
            time.sleep(BASE_SLEEP_SECONDS * (attempt + 1))
    raise RuntimeError(f'call_score failed after {MAX_RETRIES} retries: {last}')

def _dist_from_logprobs(response):
    content = getattr(response.choices[0], 'logprobs', None)
    if content is None or not getattr(content, 'content', None):
        return {}
    for token in content.content:
        dist = {}
        for alt in token.top_logprobs:
            t = alt.token.strip()
            if t.isdigit() and SCALE_LO <= int(t) <= SCALE_HI:
                dist[int(t)] = dist.get(int(t), 0.0) + math.exp(alt.logprob)
        if dist:
            z = sum(dist.values())
            return {s: p / z for s, p in dist.items()}
    return {}

def weighted_score(dist):
    return sum((s * p for s, p in dist.items())) if dist else np.nan

def argmax_score(dist):
    return max(dist, key=dist.get) if dist else np.nan

def dist_entropy(dist):
    return -sum((p * math.log(p) for p in dist.values() if p > 0)) if dist else np.nan
```

```python
AXES = [{'name': 'Accuracy', 'definition': 'Assesses whether the medical content of the protocol is factually correct, clinically appropriate, and consistent with current medical consensus and applicable clinical guidelines. Covers interpretation of symptoms, history, examination findings and test results; correctness of the diagnostic conclusion; appropriateness of recommended investigations and treatment, including drug selection, dose, route, duration, contraindications and interactions. This axis also covers recommending investigations, medications or interventions without clinical indication. It judges content correctness only: whether required content is present at all is Completeness, whether the conclusion is properly supported is Reasoning, whether the triage level is right is Urgency.'}, {'name': 'Completeness', 'definition': 'Assesses whether the protocol includes all clinically necessary components for safe and appropriate management of the case, across the protocol fields. A protocol that is accurate but omits key findings, indicated investigations, treatment components, follow-up arrangements or safety-netting instructions is deficient on this axis even if everything it does contain is correct. It judges presence or absence of required content; the correctness of the content that is present is Accuracy.'}, {'name': 'Reasoning', 'definition': 'Assesses whether the protocol integrates the patient data into a logically coherent and clinically justified assessment: the stated conclusion and plan must be supported by the findings actually cited, competing explanations must be handled consistently, and protocol fields must not contradict each other. A correct conclusion reached without supporting rationale is deficient on this axis; a well-reasoned conditional assessment under genuinely insufficient information is not penalized.'}, {'name': 'Urgency', 'definition': "Assesses whether the protocol correctly determines the urgency of the patient's condition and recommends an appropriate level and timeframe of care: recognition of emergency and time-critical conditions, correct escalation including conditional escalation pending missing information, and avoidance of both under-triage and unjustified over-triage."}, {'name': 'Uncertainty', 'definition': 'Assesses whether the protocol recognizes the limits of the available information and responds to them appropriately: epistemic calibration, maintaining a differential where the data do not permit a single diagnosis, marking provisional conclusions as provisional, not presenting assumptions as established facts; and information seeking, identifying which clinically informative data are missing and specifying the additional history, examination or investigations needed to resolve the uncertainty.'}, {'name': 'Communication', 'definition': 'Assesses whether the protocol communicates in a way that is appropriate for its audience and clinically professional: precise professional terminology, empathetic and non-stigmatizing tone, and clear expectations about illness duration, warning signs and next steps. This axis does not grade formatting or length; it grades audience fit, tone and stigma, which are content properties.'}]
TITLE = 'G-Eval (оси рубрики): new Delphi vs ready matched baselines, noroles'
PNG_NAME = 'geval_rubric_axes_framework_only.png'
```

```python
_STEPS_SYSTEM = 'You are designing an evaluation protocol for a clinical NLG task.\nGiven one evaluation metric, you write the concrete steps a rater must follow to apply it.'

def build_steps_prompt(axis):
    return f"You will be given one clinical protocol written in response to a clinical case.\nThe protocol has the fields Age, Sex, Complaint, Analysis, Required diagnostic equipment/tests,\nConclusion, Emergency type, Subclinical specialization, Recommendations, Notes.\n\nYour task is to write the evaluation steps that a rater should follow to rate the protocol\non one metric.\n\nEvaluation Criteria:\n\n{axis['name']} ({SCALE_LO}-{SCALE_HI}) - {axis['definition']}\n\nWrite 3 to 5 concrete evaluation steps. Each step must be one sentence and must be checkable\nagainst the case and the protocol. Output the numbered steps only, no preamble."
if STEPS_JSON.exists():
    METRICS = json.loads(STEPS_JSON.read_text(encoding='utf-8'))
    print(f'шаги загружены из {STEPS_JSON} ({len(METRICS)} метрик) — генерация пропущена')
else:
    METRICS = []
    for axis in tqdm(AXES, desc='Auto-CoT'):
        steps = call_text(_STEPS_SYSTEM, build_steps_prompt(axis), TEMPERATURE_STEPS, MAX_TOKENS_STEPS)
        METRICS.append({'name': axis['name'], 'definition': axis['definition'], 'steps': steps})
    STEPS_JSON.write_text(json.dumps(METRICS, ensure_ascii=False, indent=1), encoding='utf-8')
    print(f'шаги сгенерированы и заморожены в {STEPS_JSON}')
for m in METRICS:
    print('\n===', m['name'], '===')
    print(m['steps'])
```

```python
def build_geval_prompt(metric, case_text, protocol):
    return f"You will be given one clinical protocol written in response to a clinical case.\n\nYour task is to rate the protocol on one metric.\n\nPlease make sure you read and understand these instructions carefully. Please keep this\ndocument open while reviewing, and refer to it as needed.\n\nEvaluation Criteria:\n\n{metric['name']} ({SCALE_LO}-{SCALE_HI}) - {metric['definition']}\n\nEvaluation Steps:\n\n{metric['steps']}\n\nClinical case:\n<<<CASE>>>\n{case_text}\n<<<END CASE>>>\n\nProtocol:\n<<<PROTOCOL>>>\n{protocol}\n<<<END PROTOCOL>>>\n\nEvaluation Form (scores ONLY):\n\n- {metric['name']}:"
```

```python
if RUN:
    records = []
    todo = [(row, m) for _, row in protocols.iterrows() for m in METRICS]
    for row, metric in tqdm(todo, desc='G-Eval'):
        prompt = build_geval_prompt(metric, row['case_text'], row['protocol'])
        out = call_score(prompt)
        dist = out['dist']
        records.append({JOIN_KEY: row[JOIN_KEY], 'arm': row['arm'], 'metric': metric['name'], 'score': weighted_score(dist), 'argmax': argmax_score(dist), 'entropy': dist_entropy(dist), 'n_support': len(dist), 'words': row['words'], 'raw': out['text'], 'dist_json': json.dumps({str(k): round(v, 6) for k, v in sorted(dist.items())})})
    framework_scores = pd.DataFrame(records)
    baseline_scores = read_csv(SUP_SCORES_CSV)
    missing = [c for c in framework_scores.columns if c not in baseline_scores.columns]
    if missing:
        raise ValueError(f'sup baseline scores: нет колонок {missing}')
    baseline_scores = baseline_scores[baseline_scores['arm'].isin(BASELINE_ARMS)].copy()
    missing_arms = sorted(set(BASELINE_ARMS) - set(baseline_scores['arm'].unique()))
    if missing_arms:
        raise ValueError(f'sup baseline scores: нет рук {missing_arms}')
    expected_metrics = {m['name'] for m in METRICS}
    if set(baseline_scores['metric'].unique()) != expected_metrics:
        raise ValueError('Метрики baseline из sup не совпадают с текущими METRICS')
    baseline_scores = baseline_scores[framework_scores.columns]
    if SAMPLE_N:
        baseline_scores = baseline_scores[baseline_scores[JOIN_KEY].isin(framework_scores[JOIN_KEY].unique())]
    scores = pd.concat([framework_scores, baseline_scores], ignore_index=True)
    if scores.duplicated([JOIN_KEY, 'arm', 'metric']).any():
        raise ValueError('В объединённых результатах есть дубли case × arm × metric')
    scores.to_csv(OUT_CSV, sep=';', index=False, encoding='utf-8-sig')
    print(f'новый Delphi: {len(framework_scores)} строк; готовые baseline из sup: {len(baseline_scores)} строк')
    print(f'сохранено: {OUT_CSV}  ({len(scores)} строк)')
else:
    print('RUN=False — прогон пропущен. Выставьте RUN=True и OPENAI_API_KEY в окружении.')
```

```python
scores = pd.read_csv(OUT_CSV, sep=';')
pivot = scores.pivot_table(index='arm', columns='metric', values='score', aggfunc='mean')
pivot['MEAN'] = pivot.mean(axis=1)
print('Средний G-Eval балл по рукам и метрикам (шкала %d-%d):' % (SCALE_LO, SCALE_HI))
display(pivot.round(3))
print('\nРазрешающая способность: вероятностный балл против «сырого» целого')
res = scores.groupby('metric').agg(mean_weighted=('score', 'mean'), mean_argmax=('argmax', 'mean'), sd_weighted=('score', 'std'), sd_argmax=('argmax', 'std'), mean_entropy=('entropy', 'mean'), frac_saturated=('score', lambda s: float((s >= SCALE_HI - 0.01).mean())))
res['sd_gain'] = res['sd_weighted'] / res['sd_argmax'].replace(0, np.nan)
display(res.round(4))
print('sd_gain >> 1 -> вероятностное взвешивание вернуло дисперсию, которой нет у целых оценок;')
print('frac_saturated -> доля протоколов, упёршихся в потолок шкалы (эффект насыщения).')
```

```python
from scipy import stats
rows = []
for base in [a for a in scores.arm.unique() if a != 'delphi']:
    for m in scores.metric.unique():
        d = scores[(scores.arm == 'delphi') & (scores.metric == m)][[JOIN_KEY, 'score', 'words']]
        b = scores[(scores.arm == base) & (scores.metric == m)][[JOIN_KEY, 'score', 'words']]
        j = d.merge(b, on=JOIN_KEY, suffixes=('_d', '_b'))
        if not len(j):
            continue
        diff = j.score_d - j.score_b
        t = stats.wilcoxon(j.score_d, j.score_b) if diff.abs().sum() > 0 else None
        r = stats.pearsonr(diff, j.words_d - j.words_b) if diff.std() > 0 else (np.nan, np.nan)
        rows.append({'baseline': base, 'metric': m, 'n': len(j), 'delphi': j.score_d.mean(), 'data': j.score_b.mean(), 'delta': diff.mean(), 'p_wilcoxon': t.pvalue if t else np.nan, 'r(delta, dwords)': r[0]})
delta = pd.DataFrame(rows)
print('Delphi - baseline по метрикам. r(delta, dwords) — связь преимущества с разницей в длине:')
display(delta.round(4))
```

```python
import matplotlib.pyplot as plt
order = list(pivot.columns.drop('MEAN'))
arms = list(pivot.index)
x = np.arange(len(order))
w = 0.8 / len(arms)
fig, ax = plt.subplots(figsize=(max(8, 1.4 * len(order)), 4.5))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')
for i, arm in enumerate(arms):
    ax.bar(x + i * w - 0.4 + w / 2, pivot.loc[arm, order].values, width=w, label=arm)
ax.set_xticks(x)
ax.set_xticklabels(order, rotation=20, ha='right', fontsize=9)
ax.set_ylim(SCALE_LO, SCALE_HI)
ax.set_ylabel(f'G-Eval score ({SCALE_LO}-{SCALE_HI})', fontsize=11)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(frameon=False, fontsize=9, ncol=len(arms))
ax.set_title(TITLE, fontsize=12, fontweight='bold', pad=10)
fig.tight_layout()
fig.savefig(PNG_NAME, dpi=170, facecolor='white')
plt.show()
```
